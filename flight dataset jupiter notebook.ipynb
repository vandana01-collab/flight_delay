```python
import matplotlib.pyplot as plt
import seaborn as sns

# Identify numerical and categorical columns
numerical_cols = df.select_dtypes(include=np.number).columns
categorical_cols = df.select_dtypes(include='object').columns

print("\n--- Univariant Analysis for Numerical Columns ---")
for col in numerical_cols:
    print(f"\nColumn: {col}")
    display(df[col].describe())
```

```python
print("\n--- Univariant Analysis for Categorical Columns ---")
for col in categorical_cols:
    print(f"\nColumn: {col}")
    display(df[col].value_counts())
```

### Sample Visualizations for Univariant Analysis

To keep the output manageable, I'll show visualizations for a few representative numerical and categorical columns. If you'd like to see plots for other specific columns, please let me know!

```python
# Example: Histogram and Box Plot for a Numerical Column (e.g., 'DepTime')
fig, axes = plt.subplots(1, 2, figsize=(16, 5))

sns.histplot(df['DepTime'], kde=True, ax=axes[0])
axes[0].set_title('Distribution of Departure Time')

sns.boxplot(y=df['DepTime'], ax=axes[1])
axes[1].set_title('Box Plot of Departure Time')

plt.tight_layout()
plt.show()
```

```python
# Example: Count Plot for a Categorical Column (e.g., 'UniqueCarrier')
plt.figure(figsize=(12, 6))
sns.countplot(y=df['UniqueCarrier'], order=df['UniqueCarrier'].value_counts().index, palette='viridis')
plt.title('Count of Flights by Unique Carrier')
plt.xlabel('Number of Flights')
plt.ylabel('Unique Carrier')
plt.show()
```

### Bivariant Analysis: Exploring Relationships Between Columns

```python
print("\n--- Correlation Matrix for Numerical Columns ---")
plt.figure(figsize=(15, 12))
sns.heatmap(df[numerical_cols].corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix of Numerical Features')
plt.show()
```

#### Example Bivariate Visualizations

I will now provide some sample visualizations for different types of column pairs:
1.  **Numerical vs. Numerical**: Scatter plot of `DepDelay` vs `ArrDelay`.
2.  **Numerical vs. Categorical**: Box plot of `ArrDelay` by `UniqueCarrier`.
3.  **Categorical vs. Categorical**: Count plot of `UniqueCarrier` vs `CancellationCode`.

```python
# Example 1: Numerical vs. Numerical (DepDelay vs. ArrDelay)
plt.figure(figsize=(10, 7))
sns.scatterplot(x='DepDelay', y='ArrDelay', data=df, alpha=0.5)
plt.title('Departure Delay vs. Arrival Delay')
plt.xlabel('Departure Delay (minutes)')
plt.ylabel('Arrival Delay (minutes)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
# Example 2: Numerical vs. Categorical (ArrDelay by UniqueCarrier)
plt.figure(figsize=(12, 8))
sns.boxplot(x='UniqueCarrier', y='ArrDelay', data=df, palette='viridis')
plt.title('Arrival Delay by Unique Carrier')
plt.xlabel('Unique Carrier')
plt.ylabel('Arrival Delay (minutes)')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
# Example 3: Categorical vs. Categorical (UniqueCarrier vs. CancellationCode)
# Filter out 'N' from CancellationCode if it's overwhelmingly common to see other codes
filtered_df = df[df['CancellationCode'] != 'N']

if not filtered_df.empty:
    plt.figure(figsize=(12, 7))
    sns.countplot(x='UniqueCarrier', hue='CancellationCode', data=filtered_df, palette='pastel')
    plt.title('Cancellations by Unique Carrier and Cancellation Code')
    plt.xlabel('Unique Carrier')
    plt.ylabel('Count')
    plt.xticks(rotation=45)
    plt.legend(title='Cancellation Code')
    plt.tight_layout()
    plt.show()
else:
    print("No non-'N' cancellation codes found for plotting.")
```

### Multivariate Analysis: Exploring Relationships Between Multiple Columns

#### Reminder: Correlation Heatmap for Numerical Features

We previously generated a correlation heatmap which is a good starting point for multivariate understanding among numerical variables:


```python
plt.figure(figsize=(15, 12))
sns.heatmap(df[numerical_cols].corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix of Numerical Features')
plt.show()
```

#### Pair Plot for Selected Numerical Variables

A pair plot creates scatter plots for every pair of numerical variables and histograms for individual variables, allowing us to see relationships and distributions simultaneously. To keep it manageable, I'll select a few relevant numerical columns.

```python
# Select a subset of numerical columns for the pair plot
selected_numerical_cols = ['DepTime', 'ArrTime', 'ActualElapsedTime', 'ArrDelay', 'DepDelay', 'Distance']

# Create a pair plot
plt.figure(figsize=(15,15)) # This is for the overall figure size, seaborn pairplot will manage subplots
sns.pairplot(df[selected_numerical_cols].sample(n=5000, random_state=42))
plt.suptitle('Pair Plot of Selected Numerical Features (Sampled 5000 rows)', y=1.02) # Adjust suptitle position
plt.show()
```

#### Multivariate Scatter Plot (Numerical vs. Numerical with Categorical Hue)

This plot shows the relationship between two numerical variables, with a third categorical variable influencing the color of the points, helping to identify groups or patterns within the data.

```python
# Example: Departure Delay vs. Arrival Delay, colored by UniqueCarrier
plt.figure(figsize=(12, 8))
sns.scatterplot(x='DepDelay', y='ArrDelay', hue='UniqueCarrier', data=df.sample(n=5000, random_state=42), alpha=0.6)
plt.title('Departure Delay vs. Arrival Delay by Unique Carrier (Sampled 5000 rows)')
plt.xlabel('Departure Delay (minutes)')
plt.ylabel('Arrival Delay (minutes)')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
